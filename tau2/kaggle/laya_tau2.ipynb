All logic lives in `run_laya_kaggle.py` inside the dataset, which imports the project's own code. Outputs land in `/kaggle/working`.

```python
!pip install -q laya==0.3.21
import torch
print('GPU:', torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'NONE - set Accelerator to GPU T4 x2')
```

```python
import glob, runpy, zipfile, os

runners = glob.glob('/kaggle/input/**/run_laya_kaggle.py', recursive=True)
if not runners:  # the zip was attached without being unpacked: unpack it ourselves
    zips = glob.glob('/kaggle/input/**/laya-tau2.zip', recursive=True)
    assert zips, 'laya-tau2 dataset not found - use Add Input to attach it'
    zipfile.ZipFile(zips[0]).extractall('/kaggle/tmp/laya-tau2')
    runners = ['/kaggle/tmp/laya-tau2/run_laya_kaggle.py']
print('running', runners[0])
runpy.run_path(runners[0], run_name='__main__')
```
