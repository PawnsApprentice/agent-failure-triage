```python
import glob, subprocess, sys, zipfile
scripts = glob.glob('/kaggle/input/**/finetune_kaggle.py', recursive=True)
if not scripts:  # dataset zip not unpacked by Kaggle: unpack it ourselves
    zips = glob.glob('/kaggle/input/**/laya-tau2.zip', recursive=True)
    assert zips, 'laya-tau2 dataset not attached'
    zipfile.ZipFile(zips[0]).extractall('/tmp/laya-tau2')
    scripts = ['/tmp/laya-tau2/finetune_kaggle.py']
print('running', scripts[0])
subprocess.run([sys.executable, scripts[0]] + ['pipeline', '--phase', 'seeds', '--setting', 'S1'], check=True)
```
